```python
from nugget.geometries.DynamicString import DynamicString
from nugget.losses.cost_loss import CostLoss

geometry = DynamicString(
    device="cpu",
    n_strings=10,
    n_cjbs=10,
    total_points=25,
)
geom_dict = geometry.initialize_points()

cost_loss = CostLoss(device="cpu")
result = cost_loss(geom_dict)

print(result["cost_loss"].item())
print(f"Estimated total configuration cost: {result['estimated_total_cost'].item():,.2f}")
```
